```python
def detect_outliers(df):
    outlier_count={}
    for column in df.select_dtypes(include='number'):
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3-Q1
        lower_bound = Q1-1.5*IQR
        upper_bound = Q3+1.5*IQR
        outliers = df[(df[column]<lower_bound )| (df[column]>upper_bound)]
        outlier_count[column] = len(outliers)
    return outlier_count
outliers_iqr = detect_outliers(df)
outlier_df = pd.DataFrame(list(outliers_iqr.items()),columns= ['Columns','Outlier_count'])
```

```python
df["Policy_Sales_Channel"] = df["Policy_Sales_Channel"].astype("str")
df["Region_Code"] = df["Region_Code"].astype("str")
```

```python
df.info()
```

```python
import math

numerical_features = df.select_dtypes(
    include=['float64', 'int64']
).columns.drop('Response')

n_cols = 5
n_rows = math.ceil(len(numerical_features) / n_cols)

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(18, 5 * n_rows),
    squeeze=False
)

for i, col in enumerate(numerical_features):
    row, col_idx = divmod(i, n_cols)

    sns.violinplot(
        x=col,
        data=df,
        ax=axes[row, col_idx]
    )

    axes[row, col_idx].set_title(f'{col} Distribution')
    axes[row, col_idx].grid(True)

# Delete unused plots
for j in range(len(numerical_features), n_rows * n_cols):
    row, col_idx = divmod(j, n_cols)
    fig.delaxes(axes[row, col_idx])

plt.tight_layout()
plt.show()
```
